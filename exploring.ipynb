```python
databases = ["Reinstatement_information_Part_2 (1).xlsx - 2020.csv", "Reinstatement_information_Part_2 (1).xlsx - 2021.csv", 'Reinstatement_information_Part_2 (1).xlsx - 2022 Jan-Nov.csv', "Currently_Suspended_information_Part_3_as_of_12082_022 (1).xlsx - Result Set 1_ 693100 Row(s).csv"]


columns_to_read = ['Description', 'Term', 'Days Sus/Rev', 'Authority', 'Issuing Agency']

# Read the first CSV file
df_concat = pd.DataFrame()

# Iterate over the rest of the CSV files and merge them with the already merged DataFrame
for i in databases:
    df_next = pd.read_csv(i, usecols=columns_to_read)
    df_concat = pd.concat([df_concat, df_next], axis=0)
```

```python
print(df_concat.shape)
print(df_concat.columns)
df_concat.head()
```

```python
d_desc = df_concat["Description"].value_counts()

props = [i/df_concat.shape[0] for i in d_desc]
print(sum(props[-172:]))
print(props[25])
```
